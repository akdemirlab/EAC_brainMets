```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import rcParams
from matplotlib.colors import ListedColormap
import os
from comut import comut
from comut import fileparsers
import palettable
import matplotlib
```

```python
#library and functions
def plot_categorical(comut, data, name, column, mapping = None):
    data = data.copy()
    data['category'] = name
    data['value'] = data[column]
    comut.add_categorical_data(data, name = name, mapping = mapping)

def plot_continuous(comut, data, name, column, mapping, cat_mapping, value_range):
    data = data.copy()
    data['category'] = name
    data['value'] = data[column]
    data['value'] = data[column].astype('float64')

    comut.add_continuous_data(data, name = name, mapping = mapping, cat_mapping = cat_mapping, value_range = value_range)
```

```python
purp_7 = palettable.cartocolors.sequential.Purp_7.mpl_colormap
vivid_10 = palettable.cartocolors.qualitative.Vivid_10.mpl_colors
balance_6 = palettable.cmocean.diverging.Balance_6.mpl_colors



#smaller plot
from matplotlib import rcParams
custom_rcParams = {
    'font.family': 'Arial',
    'font.size': 13,
    'axes.labelsize': 13,
    'legend.fontsize':9,
    'xtick.labelsize': 5,
    'ytick.labelsize': 9
}

# update rcParams
rcParams.update(custom_rcParams)

data['sample'] = data['Patient ID']
data["Group"] = data["Group"].astype("category")
data.Group = pd.Categorical(data.Group, 
                    categories=['EAC Brain Mets', 'PCAWG Primary EAC', 'Hartwig EAC Mets'],
                    ordered=True)
data.sort_values(by=['Group', 'Patient ID'], ascending=[True,True], inplace=True)

comut1 = comut.CoMut()

# Now proceed with plotting
plot_categorical(comut1, data, "Gender", 'Gender', 
                 mapping={ 
                          'M': '#6685c2', 
                          'F': '#f1b6da', }) 



#add continuous variables
cat_mapping = {'Absent': {'facecolor': 'red'}}

plot_continuous(comut1, data, 'Age', ['Age'], mapping='Purples', cat_mapping = cat_mapping, value_range = (40, 80))
plot_continuous(comut1, data, 'BMI', ['BMI'], mapping='Greens', cat_mapping = cat_mapping, value_range = (16, 46))


plot_categorical(comut1, data, "Alcohol",'Alcohol', mapping={'Ever':'#91e3fe', 'Never':'#f5f5f5'})
plot_categorical(comut1, data, "Smoking status",'Smoking status', mapping={'Current':'#9a78fe', 'Former':'#d3c4fe', 'Never':'#f5f5f5'})

plot_categorical(comut1, data, "Survival Category",'Survival Category', mapping={'LTS':'#2166AC','STS':'#292929', 'midrange-survival':'#f5f5f5'})


plot_categorical(comut1, data, "Paired scWGS",'scWGS', mapping={'Yes':'#F3BF5A','No':'#f5f5f5'})
plot_categorical(comut1, data, "Matched Primary Xenium",'Matched Primary', mapping={'Yes':'#f1b6da','No':'#f5f5f5'})

plot_categorical(comut1, data, "Xenium",'Xenium V1', mapping={'Yes':'#fee391', 'No':'#f5f5f5'})
plot_categorical(comut1, data, "Paired scWGS",'scWGS', mapping={'Yes':'#F3BF5A','No':'#f5f5f5'})

plot_categorical(comut1, data, "FISH",'FISH', mapping={'Yes':'#C86050','No':'#f5f5f5'})
plot_categorical(comut1, data, "WGS",'WGS', mapping={'Yes':'#01665e','No':'#f5f5f5'})




hspace = 0.08
wspace = 0.05


comut1.plot_comut(figsize = (14,8), x_padding = 0.08, y_padding = 0.07, tri_padding = 0.08, 
                hspace = hspace, 
                wspace= wspace)

comut1.add_unified_legend(ncol = 1)

## adding color bar for continuous data

bmi_ax = comut1.figure.add_axes([0.98, 0.35, 0.08, 0.014])

# purity ranges from 0 to 1
norm = matplotlib.colors.Normalize(vmin=16, vmax=46)

# create the colorbar with colormap used when the continuous data was added (purp_7)
bmi_colorbar = comut1.figure.colorbar(matplotlib.cm.ScalarMappable(norm=norm, cmap='Greens'),
                                                cax=bmi_ax, orientation='horizontal')

# remove tick marks and move tick labels slightly inwards. Also remove black border
bmi_colorbar.ax.tick_params(size=1)
bmi_colorbar.set_ticks([16, 31, 46])
bmi_colorbar.set_ticklabels([16, 31, 46])
bmi_colorbar.outline.set_visible(False)

# set title of colorbar to line up with other legend elements
bmi_colorbar.set_label('BMI', fontsize=8, labelpad=3)
###


# color bars must be added manually based on figure coordinates - [left, bottom, width, height]
age_ax = comut1.figure.add_axes([0.98, 0.28, 0.08, 0.014])

# purity ranges from 0 to 1
norm = matplotlib.colors.Normalize(vmin=40, vmax=80)

# create the colorbar with colormap used when the continuous data was added (purp_7)
age_colorbar = comut1.figure.colorbar(matplotlib.cm.ScalarMappable(norm=norm, cmap='Purples'),
                                                cax=age_ax, orientation='horizontal')

# remove tick marks and move tick labels slightly inwards. Also remove black border
age_colorbar.ax.tick_params(size=1)
age_colorbar.set_ticks([40, 60, 80])
age_colorbar.set_ticklabels([40, 60, 80])
age_colorbar.outline.set_visible(False)

# set title of colorbar to line up with other legend elements
age_colorbar.set_label('Age', fontsize=8, labelpad=3)


# Save the plot as a PNG file
output_file = ".png"
plt.savefig(output_file, dpi=300, bbox_inches='tight')  # Save with high resolution
print(f"Plot saved to {output_file}")
```
